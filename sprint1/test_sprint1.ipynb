```python
import torch
import torch.nn as nn
import torch.nn.functional as F

# LeNet-5 на современном PyTorch (2025)
# Тот же самый, что Ян ЛеКун написал в 1998 году, но на 15 строк вместо 500 строк на C++

class LeNet5(nn.Module):
    def __init__(self):
        super(LeNet5, self).__init__()
        
        # Свёрточный слой 1: 6 фильтров размером 5x5
        # Вход: 1 канал (ч/б изображение), Выход: 6 каналов (6 разных паттернов)
        self.conv1 = nn.Conv2d(in_channels=1, out_channels=6, kernel_size=5)
        
        # Свёрточный слой 2: 16 фильтров размером 5x5
        # Вход: 6 каналов, Выход: 16 каналов (более сложные паттерны)
        self.conv2 = nn.Conv2d(in_channels=6, out_channels=16, kernel_size=5)
        
        # Полносвязные слои (классическая нейросеть в конце)
        self.fc1 = nn.Linear(16 * 4 * 4, 120)  # 256 -> 120
        self.fc2 = nn.Linear(120, 84)          # 120 -> 84
        self.fc3 = nn.Linear(84, 10)           # 84 -> 10 (цифры 0-9)
    
    def forward(self, x):
        # Слой 1: свёртка -> активация -> pooling
        x = F.relu(self.conv1(x))              # 28x28 -> 24x24
        x = F.max_pool2d(x, 2)                 # 24x24 -> 12x12
        
        # Слой 2: свёртка -> активация -> pooling
        x = F.relu(self.conv2(x))              # 12x12 -> 8x8
        x = F.max_pool2d(x, 2)                 # 8x8 -> 4x4
        
        # "Распрямляем" в вектор для полносвязных слоёв
        x = x.view(-1, 16 * 4 * 4)             # 16 каналов по 4x4 = 256 чисел
        
        # Полносвязные слои с активацией
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.fc3(x)                        # Без активации на выходе (логиты)
        
        return x

# Создаём модель
model = LeNet5()
print("✅ LeNet-5 создан!")
print(f"📊 Архитектура:\n{model}")

# Считаем количество параметров
total_params = sum(p.numel() for p in model.parameters())
print(f"\n🔢 Всего параметров: {total_params:,}")
print(f"   (У ЛеКуна в 1998 было ~60,000, у нас ~61,000 — почти идентично!)")
```
